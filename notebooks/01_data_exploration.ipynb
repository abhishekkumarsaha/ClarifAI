```python
import re
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
```

```python
stop_words = set(stopwords.words("english"))
lemmatizer = WordNetLemmatizer()
```

```python
def clean_text(text):
    text = text.lower()
    
    text = re.sub(r"http\S+|www\S+|https\S+", "", text)
    
    text = re.sub(r"[^a-zA-Z\s]", " ", text)
    
    words = text.split()
    
    words = [
        lemmatizer.lemmatize(word)
        for word in words
        if word not in stop_words
    ]
    
    return " ".join(words)
```

```python
sample = df["content"].iloc[0]

print("ORIGINAL:\n")
print(sample[:1000])

print("\n\nCLEANED:\n")
print(clean_text(sample)[:1000])
```

```python
df["clean_content"] = df["content"].apply(clean_text)
```

```python
X = df["clean_content"]
y = df["label"]

print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf = TfidfVectorizer(
    max_features=50000,
    ngram_range=(1, 2),
    min_df=2,
    max_df=0.95,
    sublinear_tf=True
)
```

```python
X_train_tfidf = tfidf.fit_transform(X_train)

X_test_tfidf = tfidf.transform(X_test)
```

```python
print("Training TF-IDF shape:", X_train_tfidf.shape)
print("Testing TF-IDF shape:", X_test_tfidf.shape)
```

```python
from sklearn.linear_model import LogisticRegression

logistic_model = LogisticRegression(
    max_iter=1000,
    random_state=42
)

logistic_model.fit(X_train_tfidf, y_train)

print("Logistic Regression training completed!")
```

```python
y_pred_logistic = logistic_model.predict(X_test_tfidf)

print("Predictions completed!")
```

```python
from sklearn.metrics import accuracy_score, classification_report

accuracy = accuracy_score(y_test, y_pred_logistic)

print("Logistic Regression Accuracy:", accuracy)

print("\nClassification Report:")
print(classification_report(
    y_test,
    y_pred_logistic,
    target_names=["Fake", "Real"]
))
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.naive_bayes import MultinomialNB
from sklearn.svm import LinearSVC

# Logistic Regression
lr_clean = LogisticRegression(
    max_iter=1000,
    random_state=42
)
lr_clean.fit(X_train_tfidf_clean, y_train_clean)

# Naive Bayes
nb_clean = MultinomialNB()
nb_clean.fit(X_train_tfidf_clean, y_train_clean)

# Linear SVM
svm_clean = LinearSVC(
    random_state=42
)
svm_clean.fit(X_train_tfidf_clean, y_train_clean)

print("All 3 clean models trained successfully!")
```

```python
y_pred_lr_clean = lr_clean.predict(X_test_tfidf_clean)
y_pred_nb_clean = nb_clean.predict(X_test_tfidf_clean)
y_pred_svm_clean = svm_clean.predict(X_test_tfidf_clean)

print("All predictions completed!")
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
import pandas as pd

model_predictions = {
    "Logistic Regression": y_pred_lr_clean,
    "Naive Bayes": y_pred_nb_clean,
    "Linear SVM": y_pred_svm_clean
}

comparison = []

for name, predictions in model_predictions.items():
    comparison.append({
        "Model": name,
        "Accuracy": accuracy_score(y_test_clean, predictions),
        "Precision": precision_score(y_test_clean, predictions),
        "Recall": recall_score(y_test_clean, predictions),
        "F1 Score": f1_score(y_test_clean, predictions)
    })

clean_comparison_df = pd.DataFrame(comparison)

print(
    clean_comparison_df.sort_values(
        by="F1 Score",
        ascending=False
    ).to_string(index=False)
)
```

```python
import numpy as np

predictions_matrix = np.column_stack([
    y_pred_lr_clean,
    y_pred_nb_clean,
    y_pred_svm_clean
])

ensemble_prediction = (
    np.sum(predictions_matrix, axis=1) >= 2
).astype(int)

print("Ensemble predictions created!")
```

```python
ensemble_accuracy = accuracy_score(
    y_test_clean,
    ensemble_prediction
)

ensemble_precision = precision_score(
    y_test_clean,
    ensemble_prediction
)

ensemble_recall = recall_score(
    y_test_clean,
    ensemble_prediction
)

ensemble_f1 = f1_score(
    y_test_clean,
    ensemble_prediction
)

print("Ensemble Accuracy:", ensemble_accuracy)
print("Ensemble Precision:", ensemble_precision)
print("Ensemble Recall:", ensemble_recall)
print("Ensemble F1 Score:", ensemble_f1)
```

```python
ensemble_row = pd.DataFrame([{
    "Model": "Ensemble Voting",
    "Accuracy": ensemble_accuracy,
    "Precision": ensemble_precision,
    "Recall": ensemble_recall,
    "F1 Score": ensemble_f1
}])

final_comparison = pd.concat(
    [clean_comparison_df, ensemble_row],
    ignore_index=True
)

final_comparison = final_comparison.sort_values(
    by="F1 Score",
    ascending=False
)

print(final_comparison.to_string(index=False))
```

```python
from sklearn.svm import LinearSVC
from sklearn.calibration import CalibratedClassifierCV

# Base Linear SVM
svm_base = LinearSVC(
    random_state=42
)

# Calibrate the SVM
calibrated_svm = CalibratedClassifierCV(
    svm_base,
    method="sigmoid",
    cv=5
)

calibrated_svm.fit(
    X_train_tfidf_clean,
    y_train_clean
)

print("Calibrated Linear SVM training completed!")
```

```python
y_pred_calibrated = calibrated_svm.predict(
    X_test_tfidf_clean
)

y_prob_calibrated = calibrated_svm.predict_proba(
    X_test_tfidf_clean
)

print("Calibrated predictions completed!")
```

```python
from sklearn.metrics import accuracy_score, classification_report

calibrated_accuracy = accuracy_score(
    y_test_clean,
    y_pred_calibrated
)

print("Calibrated SVM Accuracy:", calibrated_accuracy)

print("\nClassification Report:")
print(
    classification_report(
        y_test_clean,
        y_pred_calibrated,
        target_names=["Fake", "Real"]
    )
)
```
